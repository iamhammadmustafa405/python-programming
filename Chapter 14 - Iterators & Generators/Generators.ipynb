# **Generators**

## **What is a Generator?**

Python generators are a simple way of creating iterators.

```python
# This Approach is so Cluttered and Old-schooled.

# Iterable -> Class
class mera_range:
  
  def __init__(self,start,end):
    self.start = start
    self.end = end

  def __iter__(self):
    return mera_range_iterator(self)

# Iterator -> Class

class mera_range_iterator:
  def __init__(self,iterable_obj):
    self.iterable = iterable_obj
  def __iter__(self):
    return self

  def __next__(self):
    if self.iterable.start >= self.iterable.end:
      raise StopIteration
    current = self.iterable.start
    self.iterable.start +=1

    return current
```

```python
# simple example
# generator func -> has yield statement instead of return statement
# generator function -> gives generator object
# every yield statement executed when the next(generator_object) is printed or looping on generator object.

def gen_demo():

  yield "first statement"
  yield "second statement"
  yield "third statement"
```

```python
gen = gen_demo() # generator object
print(next(gen)) # first yield 
```

```python
print(next(gen)) # second yield
```

```python
print(next(gen)) # third yield
```

```python
print(next(gen)) # StopIteration Error
```

```python
# using loop
gen = gen_demo() # generator object

for i in gen: # traversing/looping through gen object iterator fetch each time a yield statement and executes it
  print(i)
```

## **Making Custom Range Function Using Generators**

```python
def mera_range(start, stop): # so simple because of generators

  for i in range(start, stop):
    yield i
```

```python
for i in mera_range(23,31):
  print(i)
```

## **Practical Application of Generators**

```python
# practical example
# i have images folder -> 20 gb i want to convert images in numpy array to train the model on that images but i can't load such big data -> therfore i use the concept of iterators using generators to deal with this problem.

import os
import cv2 # download this library -> later check it

def image_data_reader(folder_path):
  for file in os.listdir(folder_path): # read each image one by one by looping
    f_array = cv2.imread(os.path.join(folder_path,file)) # convert image into numpy array
    yield f_array

```

```python
gen = image_data_reader(r"C:\Users\Dell\Pictures\Screenshots")

next(gen) # reading each numpy array of image one by one
next(gen)
```

3. Representing Infinite Streams

```python
# printing all even no in the world.
def alleven():
  n= 0
  while True:
    yield n
    n+=2

e = alleven()

next(e)
```

```python
next(e)
```

4. Chaining Generators

```python
def fibonacci_numbers(nums):
  x,y = 0,1
  for _ in range(nums):
    x,y = y,x+y
    yield x

def square(nums):
  for num in nums:
    yield num**2

print(sum(square(fibonacci_numbers(10))))
```
