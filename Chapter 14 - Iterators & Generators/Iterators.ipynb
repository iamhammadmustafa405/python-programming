# **Iterators**

## **What is Iteration?**

Iteration is a general term for taking each item of something, one after another. Any time you use a loop, explicit or implicit, to go over a group of items, that is iteration.


```python
# example
num = [6,7,8]

for i in num:
  print(i)
```

## **What is Iterator?**

An iterator is an object that allows the porgrammer to traverse through a sequence of data without having to store the entire data in the memory.

```python
# example
L= [x for x in range(1,100000)] # it stores all the data in memory at once

# for i in L:
#   print(i*2)

import sys
sys.getsizeof(L)
```

```python
# using range()
x= range(1,1000000000) # it does not load all the data in memory, it uses iterator that fetch one data at time perform operation and then delete it and moves to the next data

# for i in x:
#   print(i**2)

sys.getsizeof(x) # it size is so small as compare to the above list method
```

## **What is Iterable?**

An iterable is object, which one can iterate over.

It generates an iterator when passed to `iter()` method

```python
L= [1,2,3]
# L is an iterable

type(iter(L)) # iter(l) -> iterator of L
```

## **Understanding How For Loop Works**

```python
num = [1,2,3]

for i in num:
  print(i)
```

```python
# behind the scene of for loop working
# 1. fetch the iterator using iter() method
iter_num = iter(num)
# 2. iterate using next() function -> it gives the state of the iterator , (in which the iterator element is in the iterable element)
next(iter_num)
```

```python
next(iter_num)
```

```python
next(iter_num)
```

```python
next(iter_num) # raises StopIteration Error because elements end and no more iteration will be done.
```

## Making Our custom for loop


```python
def mera_khudka_for_loop(iterable):
  iterator = iter(iterable)

  while True:
    try:
      print(next(iterator))
    except StopIteration:
      break
```

```python
a= [1,2,3,"Hammad","Mustafa",5.6,True]
b= (1,5.00,"hello", False)
c= range(1,23)
d= {1,6,3}
e= {0:1,8:10}
```

# Let's create our own range() function

```python
class mera_range:
  
  def __init__(self,start,end):
    self.start = start
    self.end = end

  def __iter__(self):
    return mera_range_iterator(self)
```

```python
class mera_range_iterator:
  def __init__(self,iterable_obj):
    self.iterable = iterable_obj
  def __iter__(self):
    return self

  def __next__(self):
    if self.iterable.start >= self.iterable.end:
      raise StopIteration
    current = self.iterable.start
    self.iterable.start +=1

    return current
```

```python

for i in mera_range(1,21):
  print(i)
```

```python
x= mera_range(1,21)
type(x) # obj -> mera_range
```

```python
iter(x) # obj -> mera_range_iterator
```
