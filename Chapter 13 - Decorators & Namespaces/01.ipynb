# **Decorators & Namespaces**

### Namespaces

A namespace is a space that holds names (identifiers). Programmatically speaking,namespaces are dictionary of identifiers (keys) and their objects (values)

There are 4 types of namespaces

```python
# LEGB
```

Local Namespace

Enclosing Namespace

Global Namespace

Builtin Namespace

### Scope and LEGB Rule

A Scope is a textual region of a Python program where a namespace is directly acessible

The interpreter searches for a name from the inside out, looking in the local, enclosing, global, and finally the built-in scope. If the interpreter doesn't find the name in any of these locations, then Python raises a NameError exception

## **Decorators**

A decorator in python is a function that receives another function as input and adds some functionality (decoration) to it and returns it

This can happen only because python functions are 1st Class Citizens

1st class citizens are those that can be used as input, used for output, print() del and these type of other functionalities we can do with them

There are two types of decorators

Builtin Decorators like @staticmethod,@classmethod,@abstractmethod and @property etc.

User defined decorators that we programmers can create according to our needs

```python
# python has 1st class functions
# rename
def func():
  print("Hammad")

a= func # assigning func() to a
a() # calling a
```

```python
# deleting
def func():
  print("Hammad")

del func
func()
```

```python
# decorator behaviour 
def modify(func,num): # same like decorator
  return func(num)

def square(num): # receives this func as input and modify it.
  return num**2

modify(square,4)
```

```python
# simple example -> decorator

def my_decorator(func): # this function inputs the main function that we want to decorate
  def wrapper(): # this is the modified version of input function
    print("**************************")
    func()
    print("**************************")
  return wrapper # modified function returns

def hello():
  print("Hello My Name is Hammad Mustafa.")


def display():
  print("Hello Display!")


a= my_decorator(hello)

b= my_decorator(display)
a()
b()
```

```python
# better syntax -> @decorator_name then func ---> calling simple by using func name
def my_decorator(func): # this function inputs the main function that we want to decorate
  def wrapper(): # this is the modified version of input function
    print("**************************")
    func()
    print("**************************")
  return wrapper # modified function returns

@my_decorator
def hello():
  print("Hello My Name is Hammad Mustafa.")

@my_decorator
def display():
  print("Hello Display!")


# a= my_decorator(hello)

# b= my_decorator(display)
# a()
# b()

hello()

```

```python
# meaningful example
# creating a decorator that prints the execution time for every function

import time

def timer(func):
  def wrapper():
    start = time.time()
    func()
    print("Time taken by",func.__name__, "is",time.time()-start,"secs.")
  return wrapper

@timer
def hello():
  print("I am Hammad Mustafa.")
  time.sleep(3)

@timer
def display():
  print("hello how are you?")
  time.sleep(5)

hello()
display()
```

```python
import time

def timer(func):
  def wrapper(*args): # * args can handle any no. of arguments
    start = time.time()
    func(*args) # * args can handle any no. of arguments
    print("Time taken by",func.__name__, "is",time.time()-start,"secs.")
  return wrapper

@timer
def square(num):
  time.sleep(2)
  print(num**2)

@timer
def hello():
  print("HEllo")
  time.sleep(4)


square(2)
hello()
```

```python
# A big Problem
# making a decorator that will check if the datatype of the input/parameter given to a function is correct or not?
```

```python
# decorator with arguments

def sanity_check(data_type):
  def outer_wrapper(func):
    def inner_wrapper(*args):
      if type(*args) == data_type:
        func(*args)
      else:
        raise TypeError("Ye Datatype Nai Chale Ga Bahiya")
    return inner_wrapper
  return outer_wrapper


@sanity_check(int)
def cube(num):
  print(num**3)

@sanity_check(str)
def greet(name):
  print(f"Hello {name}")

cube(5)
# cube("hello") raise TypeError

greet("Hammad")
# greet(567) Raise TypeError
```
