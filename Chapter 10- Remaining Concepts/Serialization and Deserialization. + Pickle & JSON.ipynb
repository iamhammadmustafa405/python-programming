## **Serialization & Deserialization**

**Serialization**: Process of converting Python data types to JSON Format

**Deserialization**: Process of converting JSON to Python data types.

## **JSON** **(Java Script Object Notation)**


**JSON is a Universal  Text Data Format understood by all Programming Languages.**

```python
# JSON ---> same like python dictionary ---> key value pairs
```

```python
# Serialization using JSON module ---> Writing to json file
# list
import json # ---> module in python
L= [1,2,3,4,5]

with  open("demo.json","w") as f:
    # json.dump(what_you_want_to_dump,file_handler_object)
    json.dump(L,f) # adding list to the json file

```

```python
# dict
# indent(n) ---> parameter in json ---> add indentation after every key value pair
d={
    "name": "Hammad",
    "age": 19,
    "gender": "male"
}

with open("demo.json","w") as f:
    json.dump(d,f,indent= 4) # adding dictionary to the json file
```

```python
# Deserialization --> reading the JSON file
import json
with open("demo.json","r") as f:
    d= json.load(f) # json.load(file_handler_object) ---> read/load the data from the json file in the same datatype as we dump.
    print(d)
    print(type(d))
```

```python
# serialize and deserialize ---> tuple
import json
t= (1,2,3,4,5)

with open("demo.json","w") as f:
    json.dump(t,f) # but it will stores the data in list not in tuple thats the problem


with open("demo.json","r") as f:
    d= json.load(f)
    print(d)
    print(type(d)) # its type is list
# it is the weired behaviour of json with tuple 
# but we can convert later after loading the data in list and convert it into tuple using tuple()
```

```python
# serialize and deserialize ---> nested dict

import json
d= {
    "name": "Hammad",
    "marks": [23,34,45,56]
}

with open("demo.json","w") as f:
    json.dump(d,f)

with open("demo.json","r") as f:
    d= json.load(f)
    print(d)
    print(type(d))
```

## **Serializing & Deserializing Custom Objects**

```python
class Person:
    
    def __init__(self,fname,lname,age,gender):
        self.fname= fname
        self.lname = lname
        self.age = age
        self.gender = gender

# format to printed in:
# --> Hammad Mustafa age --> 19 ---> gender ---> male
```

```python
person = Person("Hammad","Mustafa",19,"male")
```

## **As a String**

```python

import json

with open("demo.json","w") as f:
    json.dump(person,f) # Error: Object of type Person is not JSON serializable

# we cannot  (serialize) dump/write the data that is in the Object of our custom class in json file using  this method.
```

```python
# instead of we have to make a  custom function (show_object()) in which:
# 1). we have to check whether the object that we are trying to serialize in json file is the instance/object of our custom class or not
# 2). We have to give a format ( using .format() of python ) that how the data will be displayed in the json file after serializing.


# isinstance(object/instance_name,class_name) ---> this function checks whether the instance/object is the instance of class or not. (give answer in True or False)



# .format() function ---> this function will give a format to our data we provide the attributes of the instance and whenever there is a {} it will display that attribute there in {}.

# Return a formatted version of the string, using substitutions from args and kwargs. The substitutions are identified by braces ('{' and '}').


def show_object(object):
    if isinstance(person,Person):
        return "{} {} age --> {} --> gender --> {}".format(person.fname,person.lname,person.age,person.gender)


```

```python
# serializing our custom object

# default parameter --> this parameter will decide how the data would be displayed in the json file after dumping,
# in this parameter we provide our own custom function show_object()

with open("demo.json","w") as f:  # Serializing
    json.dump(person,f,default=show_object)

with open("demo.json","r") as f:  # Deserializing
    d= json.load(f)
    print(d)
    print(type(d))  # here its datatype is str
```

## **As a Dict**

```python
# same but we have to give the formating in the custom object in dict
# indent attribute --> for indentation of every key value pair--> for better looking
def show_object(object):
    if isinstance(person,Person):
        return {"name":person.fname + " " + person.lname,"age":person.age,"gender":person.gender}
    
with open("demo.json","w") as f:  # Serializing
    json.dump(person,f,default=show_object,indent= 4)

with open("demo.json","r") as f:  # Deserializing
    d= json.load(f)
    print(d)
    print(type(d))  # here its datatype is dict
```

## **Pickling**

**Pickling**: Pickling is the process whereby a Python Object heirarchy is converted into a byte stream

**Unpickling**: Unpickling is the inverse operation, whereby a byte stream (from a binary file or bytes-like object) is converted back into an object hierarchy.

```python
class Person:
    
    def __init__(self,name,age):
        self.name = name
        self.age = age
    
    def display_info(self):
        print(f"Hi, My name is {self.name} and I am {self.age} years old. ")
        
```

```python
p = Person("Hammad",19)
```

```python
import pickle # ---> import the pickle module

# Serialization

with open("person.pkl","wb") as f: # --> wb --> write binary
    pickle.dump(p,f)   #  pickle.dump(what_you_want_to_dump,file_handler_object)
    # it will dump the p object converted into bytes stream in the file person.pkl
```

```python
# now if the file is open in other folder etc.

# Deserialization

with open("person.pkl","rb") as f:
    p= pickle.load(f) # pickle.load(file_handler_object)
    # it converts the bytes stream into our custom class object

p.display_info() # we can access all the object attributes in another folder etc.. power of pickle
```

## **Pickle Vs JSON**

**Pickle** lets the user to store the data in binary format. **JSON** lets the user to store data in a human-readable text format
