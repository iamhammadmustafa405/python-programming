```python
# how to handle exceptions?
# -> try except block
```

```python
with open("sample.txt","w") as f:
  f.write("Hello World!")
```

```python
# try except demo

try:
  with open("sample.txt1","r") as f: # when the line of code is using external resource alway add in try because there is a chance of exception/error
    print(f.read())
except:
  print("File Not Found")
```

```python
# Catching specific exception
# by specifying each except block for specific error and at last for any error that we have not anticipated is excepted by generic exception

try:
  k=3
  f= open("sample.txt","r")
  print(f.read())
  print(k)
  print(9/2)
  l= [1,2,3]
  print(l[100])
except FileNotFoundError:
  print("File Not Found")
except NameError:
  print("Variable is not defined.")
except ZeroDivisionError:
  print("Can't divide by 0")
except Exception as e: # it is good write generic exception at the end of the try except block
  print(e)
```

```python
# else block -> when try block has no error the control moves to the else block in which that code is written that is 100% sure that it has no exceptions/errors

try:
  f= open("sample.txt","r")
except FileNotFoundError:
  print("file nahi mili.")
except Exception:
  print("koi to masla hai.")
else:
  print(f.read())
```

```python
# finally -> no matter try has error or not it always execute
# Control 1: try -> else -> finally ( in case of  not error)
try:
  f= open("sample.txt","r")
except FileNotFoundError:
  print("file nahi mili.")
except Exception:
  print("koi to masla hai.")
else:
  print(f.read())
finally:
  print('ye print hoga hi hoga.')
```

```python
# control 2: try -> except -> finally ( in case of exception/error)
try:
  f= open("sample2.txt","r")
except FileNotFoundError:
  print("file nahi mili.")
except Exception:
  print("koi to masla hai.")
else:
  print(f.read())
finally:
  print('ye print hoga hi hoga.')
```

# Creating Custom Exception Class


```python
class MyException(Exception): # -> our own custom exception will always be the child of Exception class in order to raise the MyException Class
  def __init__(self,message):
    print(message)


class Bank:
  def __init__(self,balance):
    self.balance = balance

  def withdraw(self,amount):
    if amount < 0:
      raise MyException("amount cannot be -ve")
    if self.balance < amount:
      raise MyException("paise nahi hain tere pass")
    self.balance = self.balance - amount
obj = Bank(50000)
try:
  obj.withdraw(7000)
except Exception as e:
  pass
else:
  print(obj.balance)
```
